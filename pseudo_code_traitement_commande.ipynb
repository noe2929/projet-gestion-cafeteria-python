PSEUDO CODE TRAITEMENT COMMANDE 

Traitement d'une commande (client, heure, demandes)
(demandes = liste de couples : code produit, quantité)

ÉTAPE 1 : VÉRIFIER
  - Pour chaque produit demandé : si son code n'existe pas dans la carte, refuser avec le motif « produit inconnu ».
  - Pour chaque quantité demandée : si elle est négative ou nulle, refuser avec le motif « quantité invalide ».
  - Regrouper les quantités si un même produit apparaît deux fois.
  - Pour chaque produit :
      * si la quantité demandée dépasse le stock, refuser avec le motif « rupture de stock » ;
      * si c'est un plat et que l'heure n'est pas dans [11:30 ; 14:00[, refuser avec le motif « hors horaires ouvertures ».
  - Si un contrôle échoue : on s'arrête tout de suite, on compte le refus dans le motif correspondant, on renvoie « refusée ».
    Ni le stock, ni le solde, ni le cumul, ni le journal ne changent.

ÉTAPE 2 : CALCULER
  - Créer la commande (elle reçoit son numéro) et y ajouter les lignes.
  - Total = somme de (prix × quantité) de chaque ligne.
  - Remise A3 (gros panier) :
      * si le total est inférieur à 8,00 € : aucune remise ;
      * sinon : 10 % pour un étudiant ou un alternant, 5 % pour les autres.
  - Avantage B3 (client fidèle) :
      * si le client est un visiteur : aucun avantage ;
      * si son cumul d'achats est d'au moins 40,00 € : -5 % de remise sur ses prochaines commandes ;
      * sinon : aucun avantage.
  - Montant à payer = total × (1 − remise A3) × (1 − avantage B3), arrondi au centime.

ÉTAPE 3 : ENCAISSER
  - Un visiteur paie comptant : il peut toujours payer.
  - Un étudiant ou un alternant peut payer si son solde − montant reste ≥ 0.
  - Un membre du personnel peut payer si son solde − montant reste ≥ −10,00 €.
  - Si le client ne peut pas payer : refuser avec le motif « solde insuffisant »
    (même fonctionnement qu'à l'étape 1 : rien n'est modifié).

ÉTAPE 4 : ENREGISTRER (seulement si tout est passé)
  - Débiter le badge du montant (sauf pour un visiteur).
  - Ajouter le montant payé au cumul d'achats du client (sauf visiteur).
  - Retirer chaque quantité du stock du produit.
  - Noter le montant payé dans la commande.
  - Ranger la commande dans le journal.
  - Renvoyer « acceptée » avec le montant payé.
